# E-Commerce EDA
Use this notebook for exploratory analysis and charts that support the Power BI dashboard.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

project_dir = Path.cwd().parent if Path.cwd().name == 'python' else Path.cwd()
df = pd.read_csv(project_dir / 'data' / 'processed' / 'online_retail_clean.csv', parse_dates=['invoice_date'])
df.head()
```

```python
df.info()
df[['quantity', 'unit_price', 'line_revenue']].describe()
```

```python
monthly = df.groupby('month', as_index=False)['line_revenue'].sum()
monthly
```

```python
plt.figure(figsize=(10, 5))
plt.plot(monthly['month'], monthly['line_revenue'], marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (£)')
plt.tight_layout()
plt.show()
```
